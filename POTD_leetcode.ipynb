```python
grid = [[1,-2,1],[1,-2,1],[3,-4,1]]
def maxp(grid):
  mod = 10 ** 9 + 7
  m, n  = len(grid), len(grid[0])
  mx = [[0] * n for _ in range(m)]
  mn = [[0] * n for _ in range(m)] # Corrected from range(n) to range(m)
  mx[0][0] = mn[0][0] = grid[0][0]
  for i in range(1, m):
    mx[i][0] = mn[i][0] = mx[i-1][0] * grid[i][0]
  for j in range(1, n):
    mx[0][j] = mn[0][j] = mx[0][j-1] * grid[0][j]
  for i  in range(1, m):
    for j in range(1,n):
      v = grid[i][j]
      candidates = [mx[i-1][j]* v , mn[i-1][j]* v,
                    mx[i][j-1]* v, mn[i][j-1]* v]
      mx[i][j] = max(candidates)
      mn[i][j] = min(candidates)
  print("Maximum matrix", mx)
  print("Minimum matrix",mn)
  return -1 if mx[-1][-1] < 0 else mx[-1][-1] % mod
maxp(grid)
```

```python
# 27/03/26
mat = [[1,2,1,2],[5,5,5,5],[6,3,6,3]]
k = 2
def f(mat, k):
  def rotate_left(row, k):
    n = len(row)
    k = k % n  # if k  = n  mean No rotation
    return row[k:] + row[:k]
  def rotate_right(row, k):
    n = len(row)
    k = k % n
    return row[-k:] + row[:-k]
  orginal = [row[:] for row in mat]
  for i , row in enumerate(mat):
    if i % 2 == 0:
      rotated = rotate_left(row, k)
    else:
      rotated = rotate_right(row, k)
    if rotated == orginal[i]:
      continue
    else:
      return False
  return True

f(mat, k)
```

```python
from re import M
#04/05/26
commands = [4,-1,3]
obstacles = []
def robotSim(commands, obstacles):
    directions = [(0,1), (1,0), (0,-1), (-1,0)]  # N, E, S, W
    dir_idx = 0
    x, y = 0, 0
    max_dist = 0
    obstacle_set = set(map(tuple, obstacles))

    for cmd in commands:
        if cmd == -2:
          dir_idx = (dir_idx - 1)% 4
            # turn left — what happens to dir_idx?
        elif cmd == -1:
          dir_idx = (dir_idx + 1)% 4
            # turn right — what happens to dir_idx?
        else:
            dx, dy = directions[dir_idx]
            for _ in range(cmd):
              nx , ny = x + dx, y + dy
              if (nx, ny) not in obstacle_set:
                x ,y = nx, ny
                max_dist = max(max_dist, x*x + y*y)
              else:
                break
    return max_dist
robotSim(commands, obstacles)
```

```python
#08/04/2026
nums = [2,3,1,5,4]
queries = [[1,4,2,3],[0,2,1,2]]
def f(nums, queries):
  for q in queries:
    idx = q[0]
    while idx <= q[1]:
      nums[idx] = (nums[idx] * q[-1]) % (10**9 + 7)
      idx += q[2]
  result = 0
  for num in nums:
    result ^= num
  return result
f(nums, queries)
```
